**Experiment 2.1:** Write a Python program to perform basic text preprocessing on a given data. The program should:

- Convert text to lowercase. (Also display how many uppercase?)
- Remove punctuation. (Also display how many different types of punctuations?)
- Remove numbers. (Also display how many numbers?)
- Remove extra whitespace. (How much extra whitespace is present?)
- Display the cleaned text.

```python
import re
import string

def basic_preprocessing(text):
    print("=" * 60)
    print("ORIGINAL TEXT")
    print("=" * 60)
    print(text)

    # 1. Count uppercase letters BEFORE converting
    uppercase_count = sum(1 for ch in text if ch.isupper())
    print(f"\nNumber of uppercase letters: {uppercase_count}")

    # Convert to lowercase
    text_lower = text.lower()

    # 2. Count punctuation types BEFORE removing
    punctuations_found = [ch for ch in text if ch in string.punctuation]
    unique_punctuations = set(punctuations_found)
    print(f"Total punctuation marks: {len(punctuations_found)}")
    print(f"Different types of punctuation used: {len(unique_punctuations)}")
    print(f"Punctuation types: {unique_punctuations}")

    # Remove punctuation
    text_no_punct = text_lower.translate(str.maketrans('', '', string.punctuation))

    # 3. Count numbers BEFORE removing
    numbers_found = re.findall(r'\d+', text_no_punct)
    print(f"\nNumbers found: {numbers_found}")
    print(f"Count of numbers: {len(numbers_found)}")

    # Remove numbers
    text_no_numbers = re.sub(r'\d+', '', text_no_punct)

    # 4. Count extra whitespace BEFORE removing
    extra_ws_matches = re.findall(r'\s{2,}', text_no_numbers)
    total_extra_ws_chars = sum(len(m) - 1 for m in extra_ws_matches)
    print(f"\nExtra whitespace occurrences: {len(extra_ws_matches)}")
    print(f"Total extra whitespace characters: {total_extra_ws_chars}")

    # Remove extra whitespace
    cleaned_text = re.sub(r'\s+', ' ', text_no_numbers).strip()

    print("\n" + "=" * 60)
    print("CLEANED TEXT")
    print("=" * 60)
    print(cleaned_text)

    return cleaned_text

cleaned_output_2_1 = basic_preprocessing(text_2_1)
```

**Experiment 2.2:** Write a Python program to perform word tokenization and sentence tokenization on a given text. Display the resulting words and sentences separately.

(Three separate code: using NLTK, using spaCy and without using NLTK/spacy)

### 2.2 (a) Using NLTK

```python
from nltk.tokenize import word_tokenize, sent_tokenize

sentences_nltk = sent_tokenize(text_2_2)
words_nltk = word_tokenize(text_2_2)

print(f"Number of sentences: {len(sentences_nltk)}")
print("Sentences:")
for i, s in enumerate(sentences_nltk, 1):
    print(f"{i}. {s}")

print(f"\nNumber of words: {len(words_nltk)}")
print("Words:")
print(words_nltk)
```

### 2.2 (b) Using spaCy

```python
import spacy

nlp = spacy.load("en_core_web_sm")
doc = nlp(text_2_2)

sentences_spacy = [sent.text.strip() for sent in doc.sents]
words_spacy = [token.text for token in doc if not token.is_space]

print(f"Number of sentences: {len(sentences_spacy)}")
print("Sentences:")
for i, s in enumerate(sentences_spacy, 1):
    print(f"{i}. {s}")

print(f"\nNumber of words: {len(words_spacy)}")
print("Words:")
print(words_spacy)
```

### 2.2 (c) Without using NLTK/spaCy

```python
import re

sentences_manual = re.split(r'(?<=[.!?])\s+', text_2_2.strip())
sentences_manual = [s for s in sentences_manual if s]

words_manual = re.findall(r'\b\w+\b', text_2_2)

print(f"Number of sentences: {len(sentences_manual)}")
print("Sentences:")
for i, s in enumerate(sentences_manual, 1):
    print(f"{i}. {s}")

print(f"\nNumber of words: {len(words_manual)}")
print("Words:")
print(words_manual)
```

**Experiment 2.3:** Write a Python program to tokenize a text and remove commonly occurring stop words such as *is, the, a, an, in, of, and, to,* etc. Display the original and filtered tokens. (Also, display Stop Words Separately).

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

stop_words = set(stopwords.words('english'))

original_tokens = word_tokenize(text_2_3)

found_stop_words = [w for w in original_tokens if w.lower() in stop_words]
filtered_tokens = [w for w in original_tokens if w.lower() not in stop_words]

print(f"Original Tokens ({len(original_tokens)}):")
print(original_tokens)

print(f"\nStop Words Found in Text ({len(found_stop_words)}):")
print(found_stop_words)

print(f"\nFiltered Tokens (Stop Words Removed) ({len(filtered_tokens)}):")
print(filtered_tokens)

print(f"\nUnique Stop Words Removed ({len(set(found_stop_words))}):")
print(set(found_stop_words))
```
